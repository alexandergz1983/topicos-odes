# 1. Cinética de Michaelis-Menten (Metabolismo)

Este script muestra cómo cambia la velocidad de una reacción enzimática a medida que aumentas la concentración de sustrato, permitiéndote alterar la velocidad máxima y la afinidad de la enzima.

Esta gráfica muestra cómo cambia la velocidad de producción de un compuesto químico dentro de la célula a medida que le agregas más sustrato (alimento o insumo). 

**Al mover el control Vmax (Velocidad Máxima):** Estás alterando la cantidad de enzimas disponibles o su eficiencia pura. Si subes la Vmax, la curva entera se desplaza hacia arriba porque la célula tiene más "máquinas" trabajando en paralelo para procesar el sustrato. 

**Al mover el control Km (Afinidad):** Cambias qué tan "atractiva" es la enzima para el sustrato. 

**Un Km bajo significa una afinidad altísima:** la curva sube de golpe muy rápido al principio; la enzima se satura casi de inmediato con poquísimo sustrato. 

**Un Km alto genera una curva más acostada:** se necesita mucho más sustrato para alcanzar la mitad de la velocidad máxima (\(V_{max}/2\)).

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Función que calcula la velocidad de Michaelis-Menten
def plot_michaelis_menten(Vmax, Km):
    # Generar concentraciones de Sustrato [S] de 0 a 50
    S = np.linspace(0, 50, 500)
    v = (Vmax * S) / (Km + S)
    
    plt.figure(figsize=(8, 5))
    plt.plot(S, v, lw=2.5, color='teal', label=f'Vmax={Vmax}, Km={Km}')
    plt.axhline(y=Vmax, color='r', linestyle='--', alpha=0.6, label='Vmax (Saturación)')
    plt.axhline(y=Vmax/2, color='gray', linestyle=':', alpha=0.6, label='Vmax / 2')
    plt.axvline(x=Km, color='gray', linestyle=':', alpha=0.6)
    
    plt.title('Cinética de Michaelis-Menten', fontsize=14, fontweight='bold')
    plt.xlabel('Concentración de Sustrato [S]', fontsize=12)
    plt.ylabel('Velocidad de Reacción (v)', fontsize=12)
    plt.xlim(0, 50)
    plt.ylim(0, max(Vmax + 2, 12))
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.show()

# Controles interactivos
interact(plot_michaelis_menten, 
         Vmax=FloatSlider(value=10.0, min=1.0, max=20.0, step=0.5, description='Vmax:'),
         Km=FloatSlider(value=5.0, min=0.5, max=15.0, step=0.5, description='Km (Afinidad):'))
```

# 2 Cinética de Michaelis-Menten aplicada a fármacos en inmunoterapia

**Ejemplo:** eliminación saturable de un anticuerpo monoclonal

Este script adapta la cinética de Michaelis-Menten a un escenario farmacológico en inmunoterapia. En lugar de pensar en una enzima que procesa un sustrato metabólico, podemos pensar en un anticuerpo monoclonal usado como fármaco inmunoterapéutico, por ejemplo:

 - anti-PD-1, como nivolumab o pembrolizumab,
 - anti-CTLA-4, como ipilimumab,
 - anti-CD20, como rituximab,
 - u otros anticuerpos con eliminación influenciada por su blanco celular.

En muchos anticuerpos monoclonales, una parte de la eliminación del fármaco puede ser saturable, porque depende de la unión a un receptor o antígeno específico. Cuando el fármaco se une a su blanco, el complejo puede ser internalizado y degradado. Este proceso se conoce de forma general como target-mediated drug disposition.

**En este ejemplo:**

 - la concentración de sustrato [S] se reemplaza por la concentración plasmática del fármaco [C],
 - la velocidad de reacción v se reemplaza por la velocidad de eliminación saturable del fármaco,
 - Vmax representa la capacidad máxima de eliminación,
 - Km representa la concentración a la cual la velocidad de eliminación alcanza la mitad de Vmax.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Función que calcula la velocidad de eliminación saturable
def plot_michaelis_menten_inmunoterapia(Vmax, Km):
    # Concentración plasmática del fármaco inmunoterapéutico
    C = np.linspace(0, 50, 500)
    
    # Velocidad de eliminación mediada por receptor/target
    v = (Vmax * C) / (Km + C)
    
    plt.figure(figsize=(8, 5))
    plt.plot(C, v, lw=2.5, color='teal',
             label=f'Vmax={Vmax}, Km={Km}')
    
    plt.axhline(y=Vmax, color='r', linestyle='--', alpha=0.6,
                label='Vmax (saturación)')
    plt.axhline(y=Vmax/2, color='gray', linestyle=':', alpha=0.6,
                label='Vmax / 2')
    plt.axvline(x=Km, color='gray', linestyle=':', alpha=0.6)
    
    plt.title('Eliminación saturable de un anticuerpo monoclonal',
              fontsize=14, fontweight='bold')
    plt.xlabel('Concentración plasmática del fármaco [C]', fontsize=12)
    plt.ylabel('Velocidad de eliminación mediada por receptor', fontsize=12)
    
    plt.xlim(0, 50)
    plt.ylim(0, max(Vmax + 2, 12))
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.show()

# Controles interactivos
interact(plot_michaelis_menten_inmunoterapia,
         Vmax=FloatSlider(value=10.0, min=1.0, max=20.0, step=0.5,
                          description='Vmax:'),
         Km=FloatSlider(value=5.0, min=0.5, max=15.0, step=0.5,
                        description='Km:'))
```

Interpretación de los parámetros

**Al mover Vmax**

En el modelo original de Michaelis-Menten, Vmax representa la velocidad máxima de una enzima.
En este ejemplo inmunoterapéutico, Vmax puede interpretarse como la capacidad máxima del organismo para eliminar el anticuerpo a través de su blanco biológico.

**Un Vmax alto puede representar:**

 - alta expresión del receptor diana,
 - alta carga de células que expresan el antígeno,
 - rápida internalización del complejo fármaco-receptor,
 - alta degradación del fármaco,
 - mayor eliminación mediada por target.

**Interpretación clínica**

Si Vmax aumenta, la curva se desplaza hacia arriba.

Eso significa que, para una misma concentración de fármaco, la velocidad de eliminación es mayor.

En términos prácticos:

Un fármaco con un componente de eliminación saturable muy potente puede requerir dosis más altas o intervalos de dosificación más cortos para mantener concentraciones terapéuticas.

Por ejemplo, en algunos anticuerpos usados contra células B, como anti-CD20, la cantidad de células diana puede influir en la eliminación del fármaco. Si hay muchas células diana, puede haber mayor consumo inicial del fármaco. Cuando esas células disminuyen, la eliminación mediada por target también puede reducirse.

**Al mover Km**

En Michaelis-Menten, Km es la concentración de sustrato necesaria para alcanzar la mitad de la velocidad máxima.

**Km bajo**

Un Km bajo significa que el sistema de eliminación alcanza la mitad de su capacidad máxima con poca concentración de fármaco.

La curva sube rápidamente al principio.

**Interpretación:**

El sistema de eliminación tiene alta afinidad aparente o alta eficiencia a bajas concentraciones del fármaco.
En inmunoterapia, esto puede significar que incluso concentraciones bajas del anticuerpo son capturadas eficientemente por el receptor o la célula diana.

**Km alto**

Un Km alto significa que se necesita mucha más concentración de fármaco para alcanzar la mitad de Vmax.

**Interpretación:**

Se necesita más fármaco para saturar significativamente la vía de eliminación.


**Interpretación clínica en inmunoterapia**

En muchos fármacos biológicos, especialmente anticuerpos monoclonales, la eliminación no siempre es perfectamente lineal.
Cuando existe eliminación saturable mediada por receptor o target, puede ocurrir que:

Dosis pequeñas → se eliminan relativamente rápido
Dosis mayores → saturan la vía → la exposición aumenta más de lo esperado

**¿Por qué importa esto en inmunoterapia?**

Porque muchos fármacos inmunoterapéuticos no son moléculas pequeñas que se metabolizan principalmente por enzimas hepáticas clásicas.
Los anticuerpos monoclonales pueden eliminarse por:

 - proteólisis inespecífica,
 - reciclaje por FcRn,
 - unión al receptor diana,
 - internalización del complejo fármaco-receptor,
 - degradación intracelular,
 - consumo por alta carga de antígeno,
 - formación de anticuerpos antifármaco en algunos casos.

Cuando la eliminación depende del target, la farmacocinética puede volverse no lineal.
  
Por eso, una curva tipo Michaelis-Menten puede ayudar a entender por qué:

 - aumentar la dosis no siempre aumenta la velocidad de eliminación de forma proporcional,
 - bajas dosis pueden tener una exposición menor de la esperada,
 - altas dosis pueden saturar la eliminación,
 - la carga tumoral o la expresión del antígeno pueden afectar la exposición al fármaco.
