# 1. Modelo de Lotka-Volterra (Microbiomas y Ecología Celular)

Este script integra las ecuaciones diferenciales en el tiempo empleando scipy. Permite ver la oscilación interactiva de dos especies concurrentes en un mismo ecosistema celular.

Representa un sistema dinámico cerrado donde dos especies dependen la una de la otra. Verás ciclos u oscilaciones infinitas que demuestran el equilibrio natural de los sistemas.

**Al mover el control alpha (α) (Nacimiento de X):** Si aumentas el crecimiento de las bacterias tipo X (presas), estas se multiplican rápidamente (línea azul). Esto genera abundante comida para el depredador Y (línea naranja), cuya población se dispara poco después.

**Al mover el control beta (β) (Muerte de X por Y):** Modula el impacto de la interacción. Un valor muy alto significa que el "depredador" destruye con demasiada facilidad a la "presa", lo que puede causar colapsos drásticos y picos de población muy agresivos.

**El desfase temporal:** Nota que el pico de la población naranja (Y) siempre ocurre un poco después del pico de la línea azul (X). Matemáticamente, esto es el reflejo del retardo en la transferencia de biomasa de una especie a otra.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from ipywidgets import interact, FloatSlider

# Sistema de ecuaciones diferenciales
def lotka_volterra(y, t, alpha, beta, delta, gamma):
    x, y_pob = y
    dxdt = x * (alpha - beta * y_pob)
    dydt = y_pob * (delta * x - gamma)
    return [dxdt, dydt]

def plot_lotka_volterra(alpha, beta, delta, gamma):
    t = np.linspace(0, 50, 1000)
    y0 = [10.0, 5.0] # Condiciones iniciales: Población X=10, Población Y=5
    
    # Resolver EDO
    sol = odeint(lotka_volterra, y0, t, args=(alpha, beta, delta, gamma))
    
    plt.figure(figsize=(10, 5))
    plt.plot(t, sol[:, 0], lw=2, color='navy', label='Especie X (ej. Presa / Bacteria A)')
    plt.plot(t, sol[:, 1], lw=2, color='orange', label='Especie Y (ej. Depredador / Bacteria B)')
    
    plt.title('Dinámica de Poblaciones (Lotka-Volterra)', fontsize=14, fontweight='bold')
    plt.xlabel('Tiempo', fontsize=12)
    plt.ylabel('Tamaño de la Población', fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='upper right')
    plt.ylim(0, max(np.max(sol), 20))
    plt.show()

# Controles interactivos
interact(plot_lotka_volterra, 
         alpha=FloatSlider(value=1.1, min=0.1, max=3.0, step=0.1, description='Crecim. X (α):'),
         beta=FloatSlider(value=0.4, min=0.05, max=1.5, step=0.05, description='Muerte X (β):'),
         delta=FloatSlider(value=0.1, min=0.01, max=1.0, step=0.05, description='Eficiencia Y (δ):'),
         gamma=FloatSlider(value=0.4, min=0.1, max=2.0, step=0.1, description='Muerte Y (γ):'))
```

# 2. Modelo de Lotka-Volterra (Inmunoterapia CAR-T y Dinámica Tumor-Inmune)

Este script adapta las clásicas ecuaciones de depredador-presa al campo de la inmunooncología. En este escenario, las células cancerosas actúan como la "presa" que intenta proliferar, mientras que las células CAR-T (linfocitos T modificados genéticamente) actúan como el "depredador" diseñado para buscar y destruir al tumor.

El modelo ilustra la "carrera armamentista" biológica entre la expansión del cáncer y la respuesta del sistema inmunológico adaptativo tras la infusión de la terapia celular.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from ipywidgets import interact, FloatSlider

# Sistema de ecuaciones diferenciales
def car_t_dynamics(y, t, alpha, beta, delta, gamma):
    tumor, car_t = y
    # Ecuación del Tumor (Presa): Crece exponencialmente, muere al contacto con CAR-T
    dtumor_dt = tumor * (alpha - beta * car_t)
    # Ecuación CAR-T (Depredador): Se expande al destruir tumor, muere por agotamiento
    dcar_t_dt = car_t * (delta * tumor - gamma)
    return [dtumor_dt, dcar_t_dt]

def plot_car_t_immunotherapy(alpha, beta, delta, gamma):
    t = np.linspace(0, 50, 1000)
    # Condiciones iniciales: Carga tumoral inicial = 10, Dosis infundida de CAR-T = 2
    y0 = [10.0, 2.0] 
    
    # Resolver EDO
    sol = odeint(car_t_dynamics, y0, t, args=(alpha, beta, delta, gamma))
    
    plt.figure(figsize=(10, 5))
    plt.plot(t, sol[:, 0], lw=2, color='crimson', label='Células Tumorales (Cáncer / Presa)')
    plt.plot(t, sol[:, 1], lw=2, color='dodgerblue', label='Células CAR-T (Inmunoterapia / Depredador)')
    
    plt.title('Dinámica Tumor vs. CAR-T (Modelo Lotka-Volterra)', fontsize=14, fontweight='bold')
    plt.xlabel('Tiempo (Días / Semanas)', fontsize=12)
    plt.ylabel('Población Celular (x10⁹ células)', fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='upper right')
    plt.ylim(0, max(np.max(sol), 20))
    plt.show()

# Controles interactivos
interact(plot_car_t_immunotherapy, 
         alpha=FloatSlider(value=0.8, min=0.1, max=2.0, step=0.1, description='(α)Prolif. Tumor (α):'),
         beta=FloatSlider(value=0.4, min=0.05, max=1.5, step=0.05, description='(β)Eficacia Citotóxica (β):'),
         delta=FloatSlider(value=0.1, min=0.01, max=1.0, step=0.05, description='(δ)Expansión CAR-T (δ):'),
         gamma=FloatSlider(value=0.4, min=0.1, max=2.0, step=0.1, description='(γ)Agotamiento CAR-T (γ):'))
```

**Interpretación Clínica y Biológica de los Parámetros**

**Al mover el control alpha (α) (Tasa de Proliferación del Tumor):**
Representa la agresividad del cáncer. Si aumentas este valor, simulas tumores de rápido crecimiento (como ciertas leucemias agresivas). El tumor intentará multiplicarse rápidamente antes de que el sistema inmune pueda controlarlo. Un valor bajo representa un tumor indolente o de crecimiento lento.

**Al mover el control beta (β) (Eficacia Citotóxica / Muerte del Tumor por CAR-T):**
Modula la letalidad de la terapia. Representa qué tan bien el receptor quimérico (el "ojo" de la célula CAR-T) reconoce el antígeno del tumor y qué tan eficiente es la célula liberando toxinas letales (perforinas y granzimas). Un valor muy alto significa que la inmunoterapia es devastadora para el cáncer, aplanando rápidamente sus picos.

**Al mover el control delta (δ) (Expansión Clonal / Activación de CAR-T):**
En la inmunoterapia real, cuando las células CAR-T encuentran a su objetivo, no solo matan, sino que reciben señales biológicas para multiplicarse masivamente (expansión in vivo). Este parámetro dicta la velocidad a la que el "ejército" de células T se refuerza y replica tras el contacto con el tumor.

**Al mover el control gamma (γ) (Agotamiento / Muerte de CAR-T):**
Uno de los mayores desafíos en la inmunoterapia moderna es el "agotamiento de las células T" (T-cell exhaustion). Tras combatir al tumor por mucho tiempo en un microambiente hostil, las células CAR-T pierden su efectividad y sufren apoptosis. Un valor alto de 
γ simula este fenómeno, permitiendo que el tumor se recupere (rebrote) porque los "soldados" inmunes mueren o se apagan prematuramente.

**El desfase temporal y la Respuesta Inmunológica:**
Nota que el pico de la población de células CAR-T (línea azul) siempre ocurre un poco después del pico del tumor (línea roja). Matemáticamente, esto es el reflejo del tiempo biológico de respuesta: cuando la carga tumoral aumenta, toma tiempo para que las células CAR-T detecten los antígenos, se activen, pasen por la expansión clonal (multiplicarse) y ejerzan su función citotóxica.
Nota clínica: En este modelo matemático puro, las poblaciones oscilan perpetuamente. Sin embargo, en un paciente real, si la línea roja (tumor) cae a cero y se mantiene allí, se considera una Respuesta Completa (remisión o cura), rompiendo el ciclo porque ya no hay antígeno que sostenga a las células CAR-T, las cuales eventualmente se contraen dejando solo células de memoria inmunitaria.

**Parametros clave:**

 - **Disminuir gamma:** γ es agotamiento/muerte del CAR-T.Si reduces el agotamiento, el CAR-T persiste más.

 - **Aumentar delta:** δ es expansión del CAR-T al detectar tumor. Si aumentas la expansión, el CAR-T responde mejor.

 - **En cambio**, β principalmente cambia el equilibrio del CAR-T y la forma de las oscilaciones, no el equilibrio tumoral.

**Nota**: β alta...una citotoxicidad muy alta sin buena persistencia o expansión puede generar una caída tumoral inicial, pero después permitir un rebote del tumor porque el CAR-T no se sostuvo.
