# 2. Ecuación de Hill (Circuitos Genéticos)

Este script simula cómo la cooperatividad (\(n\)) transforma una curva biológica suave en un interruptor digital ultrasensible (tipo ON/OFF).

Esta ecuación explica cómo los genes se encienden o apagan de forma drástica, actuando como compuertas lógicas dentro de un chip biológico.

**Al mover el control n (Coeficiente de Hill / Cooperatividad):** Es el parámetro clave.Con n = 1, la respuesta es idéntica a **Michaelis-Menten:** una subida gradual y perezosa. Con n = 2 o 3, verás que la curva toma una forma de "S" (sigmoidea). Esto indica **cooperatividad:** una molécula ayuda a que la siguiente se una más rápido.Con n = 8 o 10, la curva se vuelve un escalón casi vertical en el valor de Ka. Esto es un interruptor genético digital. Bajo el umbral, el gen está totalmente apagado (0); pasando el umbral, se enciende al máximo (ON).

**Al mover el control Ka (Umbral):** Desplaza ese "escalón" o punto medio a la izquierda o derecha, dictando exactamente cuánta señal se requiere para gatillar la respuesta del gen.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

# Función para graficar la respuesta de Hill
def plot_hill(Vmax, Ka, n):
    S = np.linspace(0, 20, 500)
    # Ecuación de Hill
    v = (Vmax * (S**n)) / (Ka**n + S**n)
    
    plt.figure(figsize=(8, 5))
    plt.plot(S, v, lw=2.5, color='crimson', label=f'Coeficiente n = {n}')
    plt.axvline(x=Ka, color='gray', linestyle=':', alpha=0.7, label=f'Ka (Umbral) = {Ka}')
    
    plt.title('Ecuación de Hill (Regulación Genética / Cooperatividad)', fontsize=14, fontweight='bold')
    plt.xlabel('Concentración del Ligando / Factor Transcripción [S]', fontsize=12)
    plt.ylabel('Respuesta / Velocidad', fontsize=12)
    plt.xlim(0, 20)
    plt.ylim(0, Vmax + 0.5)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.show()

# Controles interactivos
interact(plot_hill, 
         Vmax=FloatSlider(value=5.0, min=1.0, max=10.0, step=0.5, description='Respuesta Máx:'),
         Ka=FloatSlider(value=5.0, min=1.0, max=10.0, step=0.5, description='Ka (Umbral):'),
         n=IntSlider(value=1, min=1, max=10, step=1, description='n (Cooperat.):'))
```

# 2. Ecuación de Hill aplicada a fármacos en inmunoterapia

**Ejemplo:** activación de linfocitos T por un anticuerpo biespecífico tipo anti-CD19/anti-CD3

Este script adapta la ecuación de Hill a un escenario farmacológico en inmunoterapia. En lugar de usar la concentración de un factor de transcripción, usamos la concentración de un fármaco inmunoterapéutico, por ejemplo un anticuerpo biespecífico que conecta linfocitos T con células tumorales.

En este caso, la respuesta puede interpretarse como:

 - activación de linfocitos T,
 - liberación de citocinas,
 - muerte de células tumorales,
 - expansión clonal de linfocitos T,
 - o una combinación de respuesta inmune.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

# Función para graficar la respuesta tipo Hill en inmunoterapia
def plot_hill_inmunoterapia(Emax, EC50, n):
    C = np.linspace(0, 20, 500)
    
    # Ecuación de Hill para respuesta farmacológica activadora
    response = (Emax * (C**n)) / (EC50**n + C**n)
    
    # Concentración necesaria para alcanzar 90% de la respuesta máxima
    EC90 = EC50 * (0.9 / 0.1)**(1 / n)
    
    plt.figure(figsize=(9, 5))
    plt.plot(C, response, lw=2.5, color='crimson',
             label=f'Respuesta inmune (n = {n})')
    
    plt.axvline(x=EC50, color='gray', linestyle=':', alpha=0.8,
                label=f'EC50 = {EC50:.2f}')
    
    # Solo se grafica EC90 si cae dentro del rango mostrado
    if EC90 <= 20:
        plt.axvline(x=EC90, color='seagreen', linestyle='--', alpha=0.8,
                    label=f'EC90 ≈ {EC90:.2f}')
    
    plt.axhline(y=0.5 * Emax, color='gray', linestyle='-', alpha=0.15)
    
    plt.title('Ecuación de Hill en Inmunoterapia: Concentración del fármaco vs. Respuesta',
              fontsize=14, fontweight='bold')
    plt.xlabel('Concentración del fármaco [C] (ej. ng/mL)', fontsize=12)
    plt.ylabel('Respuesta inmune (activación T / citocinas / lisis)', fontsize=12)
    
    plt.xlim(0, 20)
    plt.ylim(0, Emax + 0.5)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.show()

# Controles interactivos
interact(plot_hill_inmunoterapia,
         Emax=FloatSlider(value=100.0, min=10.0, max=100.0, step=5.0,
                          description='Emax:'),
         EC50=FloatSlider(value=5.0, min=0.5, max=15.0, step=0.5,
                          description='EC50:'),
         n=IntSlider(value=1, min=1, max=10, step=1,
                     description='n (Hill):'))
```

**Interpretación de los parámetros**

1. **Al mover Emax**

Emax representa la respuesta máxima que puede generar el sistema inmune bajo ese tratamiento.
En inmunoterapia, Emax puede interpretarse como:
 - máxima activación posible de linfocitos T,
 - máxima liberación de citocinas,
 - máxima capacidad de lisis tumoral,
 - o máxima expansión clonal de células T.

**Interpretación clínica**

Un Emax alto puede significar que el paciente tiene un sistema inmune capaz de responder con fuerza al fármaco.
Pero también puede ser riesgoso si la respuesta es demasiado intensa, porque en inmunoterapia una activación excesiva puede causar:

 - síndrome de liberación de citocinas,
 - neurotoxicidad,
 - inflamación sistémica,
 - toxicidad inmunológica.

Por otro lado, un Emax bajo puede representar:

 - linfocitos T agotados,
 - microambiente tumoral inmunosupresor,
 - baja infiltración de células T,
 - pérdida de antígeno tumoral,
 - mala expansión clonal.

**En resumen:**
Emax no solo indica potencia del fármaco, sino también la capacidad máxima del sistema inmune del paciente para responder.

2. **Al mover EC50**
   
EC50 es la concentración necesaria para alcanzar el 50% de la respuesta máxima.

En el gráfico, la línea gris vertical representa ese punto.

**Interpretación clínica**
Un fármaco con EC50 bajo es más potente, porque necesita menor concentración para producir una respuesta importante.

**Por ejemplo:**

 - EC50 = 2 ng/mL: el fármaco activa la respuesta inmune a concentraciones bajas.
 - EC50 = 10 ng/mL: se necesita más fármaco para obtener una respuesta similar.

En inmunoterapia, una EC50 baja puede ser deseable porque permite usar dosis menores. Sin embargo, también puede reducir el margen de seguridad si la respuesta se activa demasiado cerca de concentraciones terapéuticas.

**Dicho de otra forma:**

Una EC50 baja significa que el fármaco es potente, pero puede requerir un control más fino de la dosis para evitar toxicidad.

## **Al mover n, el coeficiente de Hill**

Este es el parámetro más interesante para entender el comportamiento tipo interruptor.

**Cuando n = 1**

La curva es hiperbólica, similar a una respuesta tipo Michaelis-Menten.

La respuesta aumenta de manera gradual:

más fármaco → más respuesta


pero sin un cambio brusco.

En inmunoterapia, esto representaría una activación progresiva y relativamente predecible de linfocitos T.

**Cuando n = 2 o n = 3**

La curva empieza a volverse sigmoidea, con forma de “S”.

Esto indica cooperatividad o amplificación biológica.

En el caso de inmunoterapia, esta cooperatividad puede deberse a:

 - unión simultánea del fármaco al linfocito T y a la célula tumoral,
 - formación de sinapsis inmunológica,
 - agrupamiento de receptores,
 - amplificación de señales intracelulares,
 - liberación de citocinas que activan más células inmunes,
 - expansión clonal de linfocitos T.

Aquí pequeñas variaciones de concentración cerca del umbral pueden producir cambios importantes en la respuesta.

**Cuando n = 8 o n = 10**

La curva se vuelve casi vertical cerca de EC50.

Se comporta como un interruptor:

OFF → poca respuesta inmune
ON  → respuesta inmune máxima

En inmunoterapia, esto puede representar un escenario donde:
 - por debajo de cierta concentración del fármaco, casi no hay activación de linfocitos T,
 - pero al superar el umbral, la activación inmune se dispara casi al máximo.

Esto puede ser útil para lograr una respuesta antitumoral potente, pero también puede ser peligroso.
Una respuesta demasiado abrupta puede asociarse con:
 - síndrome de liberación de citocinas,
 - fiebre intensa,
 - hipotensión,
 - inflamación sistémica,
 - toxicidad neurológica,
 - activación inmune excesiva.

Por eso, en muchos protocolos de inmunoterapia se utilizan estrategias como:
 - escalado progresivo de dosis,
 - monitoreo de citocinas,
 - uso de corticoides o tocilizumab si hay toxicidad,
 - ajustes según carga tumoral,
 - vigilancia de síntomas neurológicos.
