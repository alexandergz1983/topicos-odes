```r
getwd()

#source('/home/alex/Documentos/topicos-odes/ODEs-2/OPTIMIZACION_PARAMETRICA/grind.R') # descargado de {http://theory.bio.uu.nl/rdb/grind.html}
source('grind.R')

# ------------------------- DEFINICIÓN DEL MODELO -----------------------------
model <- function(t, state, parms) {
    with(as.list(c(state, parms)), {
        # Ecuaciones diferenciales
        dx <- alpha1 * (1 - x) - (beta1 * x * (v * y)^gamma1) / (K1 + (v * y)^gamma1)
        dy <- alpha2 * (1 - y) - (beta2 * y * x^gamma2) / (K2 + x^gamma2)
        return(list(c(dx, dy)))
    })
}

# ------------------------- PARÁMETROS Y ESTADOS INICIALES ---------------------
# Parámetros fijos
parms <- c(
    alpha1 = 1, alpha2 = 1,
    beta1 = 200, beta2 = 10,
    gamma1 = 4, gamma2 = 4,
    K1 = 30, K2 = 1,
    v = 1  # Parámetro de bifurcación
)

# Puntos iniciales para Newton-Raphson
initial_states <- list(
    low = c(x = 1, y = 0),  # Estado bajo
    mid = c(x = 0, y = 0),  # Estado medio
    hig = c(x = 0, y = 1)   # Estado alto
)

# ------------------------- ANÁLISIS DE EQUILIBRIOS ---------------------------
# Configurar espacio fase
par(pty = "s", mfrow = c(1, 1))  # Ejes cuadrados
plane(xmin = 0, xmax = 1, ymin = 0, ymax = 1, main = "Espacio Fase y Ceroclinas")

# Encontrar puntos de equilibrio
equilibria <- lapply(initial_states, function(s) newton(s, plot = FALSE))

# ------------------------- DIAGRAMAS DE BIFURCACIÓN --------------------------
# Función para generar diagramas
plot_bifurcation <- function(var) {
    continue(
        state = equilibria$hig, parms = parms, odes = model,
        x = "v", y = var, step = 0.001,
        xmin = 0, xmax = 2, ymin = 0, ymax = 1.1,
        main = paste("Bifurcación para", toupper(var)),
        col = ifelse(var == "x", "blue", "red"),
        add = FALSE
    )
    continue(equilibria$low, parms = parms, odes = model, x = "v", y = var, add = TRUE)
    continue(equilibria$mid, parms = parms, odes = model, x = "v", y = var, add = TRUE)
}

# Generar gráficos para X e Y
par(mfrow = c(1, 2))  # Dos gráficos lado a lado
plot_bifurcation("x")
plot_bifurcation("y")

# Guardar gráficos (opcional)
# dev.copy(png, "bifurcacion.png", width = 1000, height = 500)
# dev.off()
```
